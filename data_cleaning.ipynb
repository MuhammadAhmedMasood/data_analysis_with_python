```python
# replacing missing values with median

df= df.copy()

# numeric columns
for col in ['total_amount']:
    df[col] = df[col].fillna(df[col].median())
# categorical columns:
categorical_cols = ['customer_id','country','category']

df[categorical_cols] = df[categorical_cols].fillna("Unkown")

```

```python
print(df.isnull().sum()[df.isnull().sum()>0])
```

```python
# standardizing the text:

df.head(20)
```

```python
for col in ['category','country','status']:
    df[col] = df[col].str.strip().str.lower()

# map top canonical values:
category_map = {
    'electronics': 'Electronics',
    'clothing': 'Clothing',
    'home & garden':'Home & Garden',
    'home and garden':'Home & Garden',
    'sports':'Sports',
    'SPORTS':'Sports',
    'books':'Books',
    'unkown':'Unknown'
}

status_map = {
    'delivered':'Delivered',
    'pending':'Pending',
    'shipped':'Shipped',
    'cancelled':'Cancelled'
}

country_map = {
    'united states': 'United States', 'us':'United States',
    'u.s.a':'United States', 'usa':'United States',
    'united kingdom':'United Kingdom','uk':'United Kingdom','u.k.':'United Kingdom',
    'australia':'Australia','aus':'Australia',
    'canada':'Canada',
    'germany':'Germany','de': 'Germany',
    'unkown':'Unkown'
}

df['category'] = df['category'].map(category_map).fillna(df['category'])
df['staus'] = df['status'].map(status_map).fillna(df['status'])
df['country'] = df['country'].map(country_map).fillna(df['country'])
```

```python
df.head(10)
```

```python
df['unit_price'] = df['unit_price'].astype(str).str.replace(r'[$£,]','',regex=True).str.strip()
df['unit_price'] = pd.to_numeric(df['unit_price'], errors = 'coerce')
```

```python
df.head(10)
```

```python
bool_map = {
    'Yes':True, 'yes':True, 'Y':True, 'y':True,
    'TRUE':True, 'True':True, 'true':True, '1':True, 1:True,True:True,
    'No':False, 'no':False, 'N':False,'n':False,
    'False':False,'FALSE':False,'false':False,'0':False,0:False,False:False
}
```

```python
df['is_returned'] = df['is_returned'].map(bool_map).fillna(False)
```

```python
df.head(10)
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst = True, errors='coerce')
failed = df['order_date'].isnull().sum()
```
