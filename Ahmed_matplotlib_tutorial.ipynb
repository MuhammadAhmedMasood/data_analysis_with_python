```python
import matplotlib.pyplot as plt # plt is an aliase
import numpy as np
import pandas as pd

# for jupyter notebook, use this magic command to display plots inline
%matplotlib inline 

plt.style.use('default') # for better looking style - can also use seaborn or ggplot
```

```python
# method 1: pyplot approach
x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.plot(x,y)
plt.plot(y,x)
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.title('Simple Line Plot')
plt.show() # not needed for jupyetr notebook, but will be needed in .py files
```

```python
# method 2: OOP approach

fig, ax = plt.subplots() #creates a figure and axes object (the area where plot is actually made. Axis is different!) 
ax.plot(x,y)
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_title('Line Plot (Object-oriented)')
plt.show()

# Object-Oriented approach gives you more control and is better for multiple plots
```

```python
# basic line plots

x = np.linspace(0,10,100) # 100 points from 0 to 10
y = np.sin(x)

# simple line plot
plt.figure(figsize=(8,5)) # figsize = (width, height) in inches
plt.plot(x,y)
plt.xlabel('X Values')
plt.ylabel('sin(X)')
plt.title('Sine Wave')
plt.grid(True)
plt.show()
```

```python
# mutiple line on the plot

x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)
plt.figure(figsize = (12,8))
plt.plot(x,y1, label = 'sin(x)')
plt.plot(x,y2,label = 'cos(x)')
plt.plot(x,y3, label = 'sin(x) * cos(x)')
plt.xlabel("Values of X")
plt.ylabel("Y values")
plt.title("Multiple Functions")
plt.legend() # show legend
plt.grid(True, alpha = 0.3) # alpha control transparency
plt.show()
```

```python
# customizations

x = np.linspace(0,10,20)

plt.figure(figsize = (12,6))

plt.plot(x,x,'b-', label = 'Solid Line', linewidth = 2)
plt.plot(x,x+1, 'r--', label = 'Dashed line', linewidth = 2)
plt.plot(x,x+2,'g-.', label = 'Dashe-dot line', linewidth = 2)
plt.plot(x,x+3, 'm:', label = 'dotted line', linewidth = 2)
plt.plot(x, x+4, 'co-', label = 'Circles', markersize = 8)
plt.plot(x, x+5, 's-',label = 'squares', markersize = 6)
plt.plot(x, x+6, '^-', label = 'Triangles', markersize = 6)
plt.plot(x, x+7, 'y*--', label = 'Stars', markersize = 6)

plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Line Styles and Markers')
plt.legend()
plt.grid(True, alpha = 0.3)
plt.show()

# common format string: '[color][marker][line]'
# colours: b (blue), r (red), g (green), c (cyan), m (magenta), y (yellow), k (black)
# Markers: o (circle), s (square), ^ (triangle), * (star), + (plus), x (x)
# Lines: - (solid), -- (dashed), -. (dash-dot), : (dotted)
```

```python
# using color names and hex codes

x = np.linspace(0,10,100)

plt.figure(figsize = (10,6))
plt.plot(x,np.sin(x), color = 'steelblue', linewidth = 2, label = 'steelblue')
plt.plot(x, np.cos(x), color = '#FF6B6B', linewidth = 2, label = '#FF6B6B (hex)')
plt.plot(x, np.sin(x)*0.5, color = 'darkgreen', linestyle = '--',linewidth = 2, label = 'darkgreen dashed')

plt.xlabel('X Values', fontsize = 12)
plt.ylabel('Y Values', fontsize = 12)
plt.title('Custom colors and styles', fontsize = 14, fontweight = 'bold')
plt.legend(fontsize = 10)
plt.grid(True, alpha = 0.3)
plt.show()
```

```python
# scatter plots
np.random.seed(42) # every time computer pick the starting sequence from the point and the random numbers will always be same and so code will be reproducible.
x = np.random.randn(100) # randn is the standard normal that is the gaussian distribution which is centrered around 0 with bell curve thiining towards 3 and -3
y = 2 * x + np.random.randn(100) * 0.5 # linear relationship with noise

plt.figure(figsize = (10,6))
plt.scatter(x, y, alpha = 0.6) # alpha controls transparency
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('Scatter plot')
plt.grid(True, alpha = 0.3)
plt.show()
```

```python
# scatter plot with different colours and sizes
np.random.seed(42)
x = np.random.randn(100)
y = np.random.randn(100)
colors = np.random.rand(100) #[0,1) Random colors
sizes = 1000*np.random.rand(100) # Random sizes

plt.figure(figsize = (10,6))
plt.scatter(x, y, c = colors, s = sizes, alpha = 0.6, cmap = 'coolwarm')
plt.colorbar(label = 'color Intensity') # Add Colorbar
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('Scatter plot with color and size variations', fontweight = 'bold')
plt.grid(True, alpha = 0.3)
plt.show()

# popular colormaps: 'viridis', 'plasma', 'coolwarm', 'RdY1Bu', 'Set1'
```

```python
# Bar Charts:

categories = ['A', 'B', 'C','D','E']
values = [23,45,56,87,99]

plt.figure(figsize = (10,6))
plt.bar(categories, values, color = 'steelblue', edgecolor = 'black', linewidth = 1.2)
plt.xlabel('categories')
plt.ylabel('Values')
plt.title('Bar Chart', fontweight = 'bold')
plt.grid(True, alpha = 0.3, axis = 'y') # axis = y makes grid have just horizontal lines
plt.show()
```

```python
# horizontal bar chart

plt.figure(figsize = (10,6))
plt.barh(categories, values, edgecolor = 'black', color = 'coral', linewidth = 1.2)
plt.xlabel('Values')
plt.ylabel('Categories')
plt.title('Horizontal Bar Chart', fontweight = 'bold')
plt.grid(True, alpha = 0.2, axis = 'x') # vertical lines
plt.show()
```

```python
# Grouped Bar Chart

categories = ['Q1', 'Q2', 'Q3', 'Q4']
product_a = [20, 35, 30, 33]
product_b = [25,31,41,36]
product_c = [19,31,39,51]

x = np.arange(len(categories))
width = 0.25 # width of the bars

plt.figure(figsize = (10,6))

# plt.bar(x, height, width, ...) parameters:
# x - width: x-position (shift left by one bar width to position first group)
# product_a: height values for product A bars
# width: width of each bar (0.25)
# label: text for legend
# color: bar fill color (hex code)
plt.bar(x - width, product_a, width, label = "Product A", color = '#FF6B6B')
# x: center position for second group (no shift)
plt.bar(x, product_b, width, label = 'Product B', color = '#4ECDC4')
# x + width position (shift right by one bar width to position third group)
plt.bar(x + width, product_c, width, label = 'Product C', color = '#95E1D3')

plt.xlabel('Quarter')
plt.ylabel('Sales')
plt.title('Quarterly Sales Comparison')
plt.xticks(x, categories)
plt.legend()
plt.grid(True, alpha = 0.3, axis = 'y')
plt.show()
```
