## 4. Filtering (Boolean Indexing)

```python
# filter rows where Category is Electronics
df[df["Category"]=="Electronics"]
```

```python
# filter with multiple conditions (AND)
df[(df["Category"] == "Electronics") & (df["Country"] == "USA")]
```

```python
# filter with multiple conditions (OR)
df[(df["Category"] == "Electronics") | (df["Country"] == "USA")]
```

```python
df[df["Quantity"]>2] # can do <, !=, == etc etc
```

```python
# filter rows where CustomerName starts with "A"
df[df["CustomerName"].str.startswith("A")]
```

```python
# filter rows where CustomerName ends with "a"
df[df["CustomerName"].str.endswith("a")]
```

```python
# filter rows where Country is in the list
df[df["Country"].isin(["USA","Pakistan","Brazil"])]
```

```python
df[~df["Country"].isin(["USA","Pakistan","Brazil"])] # ~ reverses it
```

## 5. Modifying Data

```python
# conditionally update a value using loc
df.loc[df["CustomerName"] == "Anna Ivanova", "Product"] = "Gill"
```

```python
# verify the update
df.loc[df["CustomerName"] == "Anna Ivanova"]
```

```python
# conditionally rename a value using loc
df.loc[df["Country"] == "USA", "Country"] = "United States"
```

```python
# verify the update
df.loc[df["Country"] == "United States"]
```

```python
# .str.upper() -> makes capital
```

## 6. Removing Rows

```python
df = df.drop(39) # removes a certain row
```

```python
df
```

## 7. Handling Missing Data

```python
df.dropna() # drop any that is NA
```

```python
df.fillna({"OrderID": 0}, inplace = True) # fill all with 0. Inplace = True modifies the existing daatframe. if false, then returns the dataframe
```

## 8. Renaming Columns

```python
# rename a column
df.rename(columns = {"OrderID": "Order ID"}, inplace=True)
```

```python
df
```

## 9. Aggregating & Summarizing

```python
# count occurrences of each unique value
df["Country"].value_counts()
```

```python
# sum Price grouped by Country
df.groupby("Country")["Price"].sum()
```

```python
# sort rows by Price
df.sort_values("Price", ascending = True)
```
